## Messages

Messages are fundamental unit of context for model in langchain. They represent the input and the output of model carrying both the content and the metadata needed to represent the state of a conversation when interactive with LLM messages are object that contain;
- **Role** - Identifies the message type (example system, user)
- **Content** - Represent the actual content of the messages(like text, image, audio, documents etc)
- **Metadata** - Optional fields such as response info, messageids, and token uses.

Langchain provides standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
import os

from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = ChatGroq(model="openai/gpt-oss-120b")
```

## Message Prompt
Alternatively you can pass it in a list of messages to the model by providing a list of messages object.

Message Types:
- **System message** - Tells the model how to behave and provide context for interactions.
- **Human message** - Represents user input and interaction with the model.
- **AI Message** - Responses genrated by the model, including text content, tools calls, and metadata.
- **Tool message** - Represents the outputs of tool calls.

### System Message
A system message represent an initial set of instructions that primes the model behaviors. You can use a system message to set the tone defined the model role and stabilize guidelines for response.

### Human Message
A human message represent user input and interaction they can contain text image audio file and any other amount of multimodal content.

### AI Message
An AI message represent the output of model innocions . They can include multimodal data, tool calls, and provide specific metadata that you can later access.

### Tool Message
Four models that support tool calling AI Message can contain tool calls Toolmases are used to pass the result of a single tool execution back to the model.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a data science expert."),
    HumanMessage("Explain about RAG application in simple terms."),
]

response= model.invoke(messages)
print(response.content)
```
